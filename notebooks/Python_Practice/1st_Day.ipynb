```python
# Q3: Check whether a number is positive, negative, or zero.

User_Input = int(input("Enter a random number: "))

if User_Input > 0:
    print(f"It is a positive number: {User_Input}")
elif User_Input < 0: 
    print(f"It is a Negative number: {User_Input}") 
else:
    print(f"It is a Zero: {User_Input}")    
```

```python
# Q4: Take marks as input and print:
# A (90+)
# B (80–89)
# C (70–79)
# D (60–69)
# F (<60)

user_marks = int(input("Enter your marks: "))

if user_marks >= 90:
    print("A")
elif user_marks >= 80 and  user_marks < 90:
    print("B")
elif user_marks >= 70 and user_marks < 80:
    print("C")
elif user_marks >= 60 and user_marks < 70:
    print("D")
else:
    print("F")               


```

```python
# Q6: Find the sum of all even numbers from 1 to 100.

i = 1
sum_even_no = 0

while i <= 100:
    i += 1
    if i % 2 == 0:
        sum_even_no = sum_even_no + i
    else:
        pass

print(sum_even_no)        
```

```python
# Q7: Print the multiplication table of a given number.

multiple_no = int(input("Enter a number: "))

i = 1

while i <= 10:
    print(f"{multiple_no} x {i} = {multiple_no * i}")
    i += 1
    
```

```python
# Q8: Count how many vowels are present in a string.

# A, E, I, O, U

string = input("Enter a string: ")

count_of_vowel = 0

for s in string:
    if s.casefold()  in ("a","e","i","o","u"):
        count_of_vowel += 1
    print(s,end=" ")

print(count_of_vowel)
```

```python
string = input("Enter a string: ")

count_of_vowel = sum(1 for s in string.lower() if s in "aeiou")

print(count_of_vowel)
```

```python
# Q9: Find the largest and smallest element in a list without using max() or min().


list = [1,1,2,4,0,5,7,8]

list_1 = sorted(list)

min_no = list_1[0]

max_no = list_1[-1]

print(min_no)
print(max_no)

```

```python
numbers = [1, 1, 2, 4, 0, 5, 7, 8]

min_no = numbers[0]
max_no = numbers[0]

for num in numbers:
    if num < min_no:
        min_no = num

    if num > max_no:
        max_no = num

print("Minimum:", min_no)
print("Maximum:", max_no)
```
