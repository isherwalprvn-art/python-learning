```python
# Check the pipeline status and print the correct action message.

status = "Failed"
# Rules:
# "running"  → "Pipeline is active. No action needed."
# "failed"   → "ALERT: Pipeline failed! Restart immediately."
# "success"  → "Pipeline completed successfully."
# anything else → "Unknown status. Please check logs."

if status.casefold() == 'running':
    print("Pipeline is active. No action needed.")
elif status.casefold() == 'failed':
    print("ALERT: Pipeline failed! Restart immediately.")
elif status.casefold() == 'success':
    print('Pipeline completed successfully.')
else:
    print("Unknown status. Please check logs.")            
```

```python
# Write a function celsius_to_fahrenheit(c) and convert all values in the list.


# Expected:
# 0°C   → 32.0°F
# 20°C  → 68.0°F ...

# formula
# (celsius * 9/5) + 32

temps = [0, 20, 37, 100, -10]

def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32       # takes c, returns the converted value

for tf in temps:
    result = celsius_to_fahrenheit(tf)   # pass one value at a time
    print(f"{tf}°C  →  {result}°F")
```

```python
# Count how many times each vowel appears in the text.

text = "data engineering is a rewarding career"
# Expected:
# a: 5
# e: 4
# i: 2 ...

# Vowels
# vowels = "aeiou"          # lowercase
# vowels = "aeiouAEIOU"     # both cases

text = text.casefold()

count_a = 0
count_e = 0
count_i = 0
count_o = 0
count_u = 0

for ch in text:           # loop over each character directly
    if ch == 'a':
        count_a += 1
    elif ch == 'e':
        count_e += 1
    elif ch == 'i':
        count_i += 1
    elif ch == 'o':
        count_o += 1
    elif ch == 'u':
        count_u += 1

print(f"a: {count_a}")   # a: 5
print(f"e: {count_e}")   # e: 4
print(f"i: {count_i}")   # i: 2
print(f"o: {count_o}")   # o: 0
print(f"u: {count_u}")   # u: 0


```

```python
# Write a function is_palindrome(word) that returns True if the word reads same forwards and backwards.

words = ["radar", "python", "level", "data", "madam"]
# Expected:
# radar  → True
# python → False ...

def is_palindrome(word):
    return word == word[::-1]

for w in words:
    result = is_palindrome(w)
    print(f'{w} → {result}')


    
```

```python
def flatten(lst):
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

nested_list = [1,2,[3,4,[5,6]],7]
print(flatten(nested_list))
```
