```python
import math

class Circle:
    def __init__(self, r=1):
        self.radius = r

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

    def __str__(self):
        return (f"Circle(radius={self.radius})\n"
                f"Area      = {self.area():.2f}\n"
                f"Perimeter = {self.perimeter():.2f}")

c = Circle(7)
print(c)
```

```python
from enum import Enum

class Order_Status(Enum):      
    PENDING     = 1
    PROCESSING  = 2            
    SHIPPED     = 3
    DELIVERED   = 4
    CANCELLED   = 5

class Order:
    def __init__(self, order_id, customer_name):
        self.order_id       = order_id
        self.customer_name  = customer_name
        self.status         = Order_Status.PENDING

    def status_update(self, new_status):
        self.status = new_status          # ✅ assignment, not return

    def details(self):
        return (f"Order ID      : {self.order_id}\n"
                f"Customer Name : {self.customer_name}\n"
                f"Order Status  : {self.status.name}")  # .name gives clean text

# Testing
o = Order(101, "Parveen")
print(o.details())

o.status_update(Order_Status.SHIPPED)
print("\nAfter update:")
print(o.details())
```

```python
import heapq

class Task:
    def __init__(self):
        self.inner = self.Priority()        # ✅ create inner class object correctly

    class Priority:
        def __init__(self):
            self.heap = []                  # ✅ heap must be initialized first
            heapq.heappush(self.heap, (5, "Reorder Inventory"))
            heapq.heappush(self.heap, (4, "Prepare Report"))
            heapq.heappush(self.heap, (2, "Update Website"))
            heapq.heappush(self.heap, (1, "Update Inventory"))
            heapq.heappush(self.heap, (3, "SMS Customer"))

        def show(self):
            print("Tasks by priority:")
            temp = self.heap.copy()         # ✅ keep original heap intact
            while temp:
                priority, task = heapq.heappop(temp)
                print(f"  Priority {priority} : {task}")

# Using it
o = Task()
o.inner.show()
```

```python
import heapq

class Task:
    def __init__(self):
        self.inner = self.Priority()

    class Priority:
        def __init__(self):
            self.heap = []
            heapq.heappush(self.heap, (5, "Reorder Inventory"))
            heapq.heappush(self.heap, (4, "Prepare Report"))
            heapq.heappush(self.heap, (2, "Update Website"))
            heapq.heappush(self.heap, (1, "Update Inventory"))
            heapq.heappush(self.heap, (3, "SMS Customer"))

        def show_next(self):
            if self.heap:
                priority, task = self.heap[0]   # peek at top
                print(f"Current Task → Priority {priority} : {task}")
            else:
                print("✅ All tasks completed!")

        def complete(self):
            if self.heap:
                priority, task = heapq.heappop(self.heap)  # remove top
                print(f"✔ Completed : {task}")
                self.show_next()                           # show next one
            else:
                print("✅ No more tasks!")

        def show_all(self):
            if not self.heap:
                print("No tasks remaining.")
                return
            print("--- Remaining Tasks ---")
            temp = self.heap.copy()
            while temp:
                priority, task = heapq.heappop(temp)
                print(f"  Priority {priority} : {task}")

# Using it
o = Task()

o.inner.show_all()          # show all tasks

print("\n--- Completing Tasks One by One ---\n")
o.inner.complete()          # complete priority 1 → show priority 2
print()
o.inner.complete()          # complete priority 2 → show priority 3
print()
o.inner.complete()          # complete priority 3 → show priority 4
```
